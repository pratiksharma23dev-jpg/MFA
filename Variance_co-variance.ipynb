```python
X = [1.70,1.62,1.52,1.85,1.91,1.42]
Y = [72,64,84,80,72,70]

n = len(X)
sum_X = 0

for x in X:
    sum_X = sum_X + x

mean_X = sum_X / n

sum_Y = 0

for y in Y:
    sum_Y = sum_Y + y

mean_Y = sum_Y / n

sum_squared_X = 0

for x in X:
    sum_squared_X = sum_squared_X + (x - mean_X) ** 2

variance_X = sum_squared_X / (n - 1)

sum_squared_Y = 0

for y in Y:
    sum_squared_Y = sum_squared_Y + (y - mean_Y) ** 2

variance_Y = sum_squared_Y / (n - 1)

sum_covariance = 0

for i in range(n):
    sum_covariance = sum_covariance + (X[i] - mean_X) * (Y[i] - mean_Y)

covariance = sum_covariance / (n - 1)



covariance_matrix = [

        [variance_X, covariance],

        [covariance, variance_Y]

    ]


print("Mean X:", mean_X)
print("Mean Y:", mean_Y)
print("Variance X:", variance_X)
print("Variance Y:", variance_Y)
print("Covariance Matrix:",covariance_matrix)
```
